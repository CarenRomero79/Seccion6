1. Marco de Gobernanza

Es el conjunto de normas, roles, responsabilidades, procesos y mecanismos de control que permiten dirigir y supervisar correctamente un proyecto. Define quién toma las decisiones, cómo se controlan los recursos, cómo se gestionan los riesgos y cómo se verifica el cumplimiento de los objetivos.

2. Integridad del proyecto

Significa garantizar que el proyecto se desarrolle de manera ética, transparente, responsable y coherente con sus objetivos. Implica proteger la información, utilizar adecuadamente los recursos, cumplir las normas y evitar conflictos de interés, fraude o manipulación de resultados.

3. Flexibilidad

Es la capacidad del proyecto para adaptarse a cambios en las necesidades, recursos, tecnología, riesgos o condiciones del entorno sin perder sus objetivos principales. Un proyecto flexible puede modificar actividades, tiempos o estrategias cuando sea necesario, manteniendo el control y la calidad.

En conjunto: el marco de gobernanza establece cómo se dirige el proyecto, la integridad garantiza que se haga correctamente y de forma responsable, y la flexibilidad permite adaptarse a los cambios sin perder el rumbo.


2. Roles claves del equipo del proyecto
Patrocinador (Sponsor): es la persona o entidad que apoya y respalda el proyecto a nivel estratégico. Proporciona recursos, facilita la toma de decisiones importantes, ayuda a resolver problemas que superan al equipo y verifica que el proyecto esté alineado con los objetivos de la organización.
Product Owner: representa las necesidades del cliente y del negocio. Define y prioriza el Product Backlog, buscando que el equipo desarrolle primero las funcionalidades que generan mayor valor.
Scrum Master: facilita la aplicación de la metodología Scrum. Ayuda a eliminar obstáculos, facilita las reuniones y promueve la colaboración y la mejora continua. No es el jefe del equipo.
Desarrolladores: son los responsables de diseñar, programar, probar y construir el producto. Trabajan de manera colaborativa para entregar un incremento funcional y de calidad en cada Sprint.
Resumen
Rol	Función principal
Patrocinador	Apoya, proporciona recursos y toma decisiones estratégicas
Product Owner	Define y prioriza qué se debe desarrollar
Scrum Master	Facilita cómo trabajar con Scrum y elimina impedimentos
Desarrolladores	Construyen y entregan el producto

Flujo sencillo:
Patrocinador → respaldo estratégico
Product Owner → valor y prioridades
Scrum Master → facilitación
Desarrolladores → construcción del producto

3. centralizado vs descentralizado

La diferencia está principalmente en dónde se toman las decisiones del proyecto.

Aspecto	Centralizado	Descentralizado
Toma de decisiones	Se concentra en un líder o grupo directivo	Se distribuye entre los miembros o equipos
Control	Mayor control central	Mayor autonomía de los equipos
Velocidad	Puede ser más lento por depender de autorizaciones	Puede ser más rápido
Participación	Menor participación del equipo	Mayor participación
Responsabilidad	Principalmente en los niveles directivos	Compartida entre los equipos
Ventaja	Facilita mantener una dirección uniforme	Favorece la innovación y adaptación
Desventaja	Puede generar dependencia y burocracia	Puede generar falta de coordinación

Ejemplo en un proyecto Scrum:

Centralizado: el patrocinador o gerente toma la mayoría de las decisiones y el equipo espera sus autorizaciones.
Descentralizado: el Product Owner, Scrum Master y desarrolladores tienen autonomía dentro de sus responsabilidades para tomar decisiones y resolver problemas.

En resumen:
👉 Centralizado = más control y autoridad concentrada.
👉 Descentralizado = más autonomía y decisiones distribuidas.


4. Autogobierno ágil

El autogobierno ágil es la capacidad que tiene un equipo para organizarse, tomar decisiones y distribuir el trabajo por sí mismo, sin depender constantemente de un jefe o de autorizaciones externas.

En un equipo Scrum:

Los desarrolladores deciden cómo realizar técnicamente el trabajo.
El Product Owner decide qué funcionalidades tienen mayor prioridad.
El Scrum Master facilita el proceso y ayuda a eliminar obstáculos.
El equipo trabaja de manera colaborativa, responsable y autónoma.

Ejemplo:
Si durante un Sprint los desarrolladores encuentran un problema técnico, pueden reunirse y decidir entre ellos la mejor solución, sin esperar que un gerente les diga exactamente cómo resolverlo.

Beneficios:

Mayor autonomía.
Decisiones más rápidas.
Mayor compromiso del equipo.
Fomenta la innovación.
Facilita la adaptación a los cambios.

En resumen:
👉 Autogobierno ágil = el equipo tiene autonomía para organizarse y tomar decisiones dentro de sus responsabilidades, manteniendo alineación con los objetivos del proyecto.

5. Plan de Gobernanza

El plan de gobernanza establece cómo se va a dirigir, controlar y supervisar un proyecto. Define quién toma las decisiones, quién es responsable de cada actividad, cómo se gestionan los recursos y cómo se controlan los riesgos y cambios.

Elementos principales:

Roles y responsabilidades: define las funciones del patrocinador, Product Owner, Scrum Master y desarrolladores.
Toma de decisiones: establece quién puede tomar decisiones y cuándo se requiere aprobación.
Comunicación: determina cómo y cuándo se informa el avance del proyecto a los interesados.
Gestión de riesgos: identifica riesgos y establece acciones para prevenirlos o reducir su impacto.
Control de cambios: define cómo se solicitan, analizan y aprueban los cambios.
Seguimiento: establece indicadores para verificar el avance, calidad, costos y cumplimiento de objetivos.
Transparencia: garantiza que la información relevante del proyecto sea clara, accesible y verificable.

Ejemplo:
En un proyecto Scrum, el Product Owner prioriza las funcionalidades, el Scrum Master facilita el proceso, los desarrolladores construyen el producto y el patrocinador proporciona apoyo estratégico y recursos.

👉 En resumen: el plan de gobernanza permite que el proyecto tenga orden, responsabilidades claras, control y transparencia, sin impedir la flexibilidad y autonomía propias de un enfoque ágil.

Políticas, límites y escalamiento en la gobernanza

Estos tres elementos ayudan a definir cómo se toman decisiones y cómo se controla un proyecto.

1. Políticas

Son las reglas y principios que deben seguirse durante el proyecto.

Ejemplos:

Política de seguridad de la información.
Política de calidad.
Política de protección de datos.
Política para aprobar cambios.
Política de comunicación y reportes.

👉 Definen qué está permitido y qué debe cumplirse.

2. Límites

Establecen hasta dónde puede actuar cada persona o rol sin solicitar autorización.

Ejemplo:
El Product Owner puede cambiar la prioridad del Product Backlog, pero un cambio que aumente considerablemente el presupuesto puede requerir autorización del patrocinador.

👉 Definen el nivel de autonomía y autoridad.

3. Escalamiento

Es el proceso mediante el cual un problema o decisión se eleva a un nivel superior cuando el equipo no tiene autoridad, recursos o capacidad para resolverlo.

Ejemplo:

Desarrolladores → Scrum Master → Product Owner → Patrocinador

Si existe un problema técnico, puede resolverlo el equipo. Si afecta el alcance, presupuesto o plazo y supera sus límites, se escala al nivel correspondiente.

En resumen

Políticas = reglas 📋
Límites = hasta dónde puedo decidir 🚧
Escalamiento = a quién recurro cuando no puedo resolverlo ⬆️

6. Matriz RACI

La matriz RACI es una herramienta de gobernanza que permite definir quién hace qué dentro de un proyecto. Evita confusiones, duplicidad de tareas y falta de responsables.

RACI significa:

Letra	Significado	Función
R – Responsible	Responsable	Es quien realiza la tarea.
A – Accountable	Aprobador / Responsable final	Es quien responde por el resultado y toma la decisión final.
C – Consulted	Consultado	Persona que debe ser consultada antes de tomar una decisión.
I – Informed	Informado	Persona que debe recibir información sobre el avance o resultado.
