Un portafolio agrupa diferentes proyectos y programas que se gestionan de manera conjunta para alcanzar los objetivos estratégicos de una organización.

La relación sería:

Visión de la empresa → Estrategia → Portafolio → Programas y proyectos → Resultados y valor

La dirección ejecutiva utiliza esta información para decidir en qué proyectos invertir recursos y cuáles están más alineados con la estrategia empresarial.

Crtioterios de exito con metricas

En esta etapa se debe identificar qué necesita el negocio y convertir esa necesidad en objetivos concretos.

El proceso puede resumirse así:

Identificar necesidad comercial → Definir objetivos medibles → Establecer criterios de éxito → Seleccionar métricas

Por ejemplo, para una inmobiliaria:

Necesidad: mejorar la estrategia de venta de viviendas.

Objetivo: identificar segmentos de casas según precio y área.

Criterio de éxito: obtener tres segmentos claramente diferenciados.

Métrica: número de clusters identificados y características promedio de cada segmento.
